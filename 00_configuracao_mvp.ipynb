# 00 — Configuração do MVP

Este notebook realiza a configuração inicial do ambiente do projeto no Databricks. Nesta etapa, são verificados o catálogo e o schema ativos, criadas as camadas Bronze, Silver e Gold e configurado o Volume utilizado para armazenamento dos arquivos brutos na nuvem.

Essa estrutura servirá como base para as etapas seguintes de ingestão, tratamento, integração e análise dos dados.

```python
# Verificar os catálogos disponíveis no Databricks
spark.sql("SHOW CATALOGS").show()
```

```python
# Verificar catálogo e schema atualmente selecionados
spark.sql("""
SELECT
    current_catalog() AS catalogo_atual,
    current_schema() AS schema_atual
""").show()
```

```python
# Criar as camadas do Data Lakehouse
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.bronze")
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.silver")
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.gold")

print("Schemas Bronze, Silver e Gold criados com sucesso!")
```

```python
# Verificar os schemas existentes no catálogo workspace

spark.sql("SHOW SCHEMAS IN workspace").show()
```

```python
# Criar um Volume para armazenamento dos arquivos brutos
spark.sql("""CREATE VOLUME IF NOT EXISTS workspace.bronze.raw_files""")

print("Volume raw_files criado com sucesso!")
```
