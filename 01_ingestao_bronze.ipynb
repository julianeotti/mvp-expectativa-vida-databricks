# 01 — Ingestão de Dados na Camada Bronze

Este notebook realiza a ingestão dos arquivos WHO_life_expectancy.csv e IDH.csv, previamente armazenados no Volume do Databricks, e cria as tabelas correspondentes na camada Bronze.

Nesta etapa, os dados são preservados próximos ao formato original das fontes, sem aplicação de tratamentos ou regras de qualidade, garantindo rastreabilidade para as transformações realizadas posteriormente.

```python
# Caminhos dos arquivos armazenados no Volume
caminho_who = "/Volumes/workspace/bronze/raw_files/WHO_life_expectancy.csv"
caminho_idh = "/Volumes/workspace/bronze/raw_files/IDH.csv"
```

```python
# Leitura da base WHO em seu formato original
df_who_bronze = (
    spark.read
    .option("header", "true")
    .option("sep", ";")
    .option("encoding", "UTF-8")
    .csv(caminho_who)
)

display(df_who_bronze)
```

```python
# Verificar estrutura e tipos identificados
df_who_bronze.printSchema()
```

```python
# Leitura da base de desenvolvimento humano em seu formato original
df_idh_bronze = (
    spark.read
    .option("header", "true")
    .option("sep", ";")
    .option("encoding", "UTF-8")
    .csv(caminho_idh)
)

display(df_idh_bronze)
```

```python
df_idh_bronze.printSchema()
```

```python
df_who_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.bronze.who_life_expectancy")
```

```python
df_idh_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.bronze.idh")
```
